```python
import pandas as pd
import numpy as np


df = pd.read_csv("C:\\Users\\SHIVANGHI SINGH\\Downloads\\archive\\WA_Fn-UseC_-Telco-Customer-Churn.csv")


df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")
df.fillna({"TotalCharges": 0}, inplace=True)


df["CLV"] = df["MonthlyCharges"] * df["tenure"]
df["Churn_Flag"] = df["Churn"].apply(lambda x: 1 if x == "Yes" else 0)


df["ContractType"] = df["Contract"].replace({
    "Month-to-month": "Short-Term",
    "One year": "Medium-Term",
    "Two year": "Long-Term"
})


df.to_csv("cleaned_telco_churn.csv", index=False, encoding='utf-8')
print("✅ Cleaned dataset ready for SQL + Tableau.")
```

```python
import os
import pandas as pd
import numpy as np


df = pd.read_csv("C:\\Users\\SHIVANGHI SINGH\\Downloads\\archive\\WA_Fn-UseC_-Telco-Customer-Churn.csv")


df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")
df.fillna({"TotalCharges": 0}, inplace=True)


df["CLV"] = df["MonthlyCharges"] * df["tenure"]
df["Churn_Flag"] = df["Churn"].apply(lambda x: 1 if x == "Yes" else 0)

# Segment contract type
df["ContractType"] = df["Contract"].replace({
    "Month-to-month": "Short-Term",
    "One year": "Medium-Term",
    "Two year": "Long-Term"
})


downloads_path = os.path.join(os.path.expanduser("~"), "Downloads", "cleaned_telco_churn.csv")


df.to_csv(downloads_path, index=False, encoding='utf-8')
print(f"✅ Cleaned dataset saved directly to Downloads: {downloads_path}")
```
